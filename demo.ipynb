# MERGE: Minimal Expression-Replacement GEneralization Test for Natural Language Inference

MERGE broadly tests if models get consistent correct predictions over variants of NLI problems. These variants are designed to preserve reasoning by replacing words not curcial to the reasoning of the problem.  

Variants are obtained by:
1.  detecting shared words between premise and hypothesis;
2.  using MLMs to suggest replacements for the shared words, filtered by various crietria.

##Example
## 1 · The original pair

**Premise:** A **black dog** is swimming with a **ball** in its mouth.

**Hypothesis:** A **black dog** found a **ball** in the water and is bring it back to its owner.

**Example label: Neutral.**

## 2 · The variants

For the shared word **ball**:

**Premise:** A black dog is swimming with a **banana** in his mouth.  
**Hypothesis:** A black dog found a **banana** in the water and is bring it back to its owner.  
**Label:** Entailment

---

**Premise:** A black dog is swimming with a **bottle** in his mouth.  
**Hypothesis:** A black dog found a **bottle** in the water and is bring it back to its owner.  
**Label:** Entailment

---

**Premise:** A black dog is swimming with a **bullet** in his mouth.  
**Hypothesis:** A black dog found a **bullet** in the water and is bring it back to its owner.  
**Label:** Entailment
